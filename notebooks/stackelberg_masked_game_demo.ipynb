## Short interpretation

In this toy game, mixed commitment weakly improves over pure commitment by flattening attacker incentives across top attacks.
Under the budget limit, some expensive masks are infeasible, so the optimal mixed policy randomizes over lower-cost masks that reduce the maximum attacker payoff.
This is consistent with Stackelberg intuition: visible randomization can reduce exploitability even when no single pure action dominates.
