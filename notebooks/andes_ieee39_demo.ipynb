# ANDES IEEE 39 Example

This notebook demonstrates the packaged IEEE 39-bus ANDES example-system API. It uses package imports and does not modify `sys.path`.


```python
from masked_perturbation_model.cases import load_ieee39_case

case = load_ieee39_case()
case.summary()
```

## Build the ANDES system

The following cell requires the optional ANDES dependencies: `pip install "mpmgame[andes]"`. It loads the ANDES IEEE39 case, computes the small-signal state matrix through ANDES, and builds a lightweight LFT-style container.


```python
try:
    system = case.build_system()
    lft = case.build_lft(system_model=system)
    print(system.summary())
    print({"nstates": lft.nstates, "ninputs": lft.ninputs, "noutputs": lft.noutputs})
except ImportError as exc:
    print(exc)
```
